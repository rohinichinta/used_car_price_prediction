```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("Dataset/used_cars.csv")
df
```

```python
df.columns
```

```python
df.isna().sum()
```

```python
df.select_dtypes(include='object')
```

```python
df.select_dtypes(include="number")
```

```python
df = df.drop(columns=['Unnamed: 0'])
```

```python
df
```

```python
from sklearn.model_selection import train_test_split
```

```python
x = df.drop("price",axis=1)
y = df["price"]
```

```python
x
```

```python
y
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.33,random_state=42)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.svm import SVR
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.preprocessing import StandardScaler,OneHotEncoder
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score

```

```python
nums_col = ["year","mileage","tax","mpg","engineSize"]

cats_col = ["model","transmission","fuelType","Make"]
```

```python
nums_pipeline = StandardScaler()
nums_pipeline
```

```python
cats_pipeline = OneHotEncoder(handle_unknown="ignore")
cats_pipeline
```

```python
preprocessor = ColumnTransformer([
    ("nums",nums_pipeline,nums_col),
    ("cats",cats_pipeline,cats_col)

])
```

```python
preprocessor
```

```python
linear_reg_pipeline = Pipeline([
    ("preprocessor",preprocessor),
    ("model",LinearRegression())
])
```

```python
linear_reg_pipeline.fit(x_train,y_train)
```

```python
lr_predict = linear_reg_pipeline.predict(x_test)
lr_predict
```

```python

print("mae: ",mean_absolute_error(y_test,lr_predict))
print("mse: ",mean_squared_error(y_test,lr_predict))
print("r2: ",r2_score(y_test,lr_predict))
```

```python
svm_pipeline = Pipeline([
    ("preprocessor",preprocessor),
    ("model",SVR())

])
```

```python
svm_pipeline.fit(x_train,y_train)
```

```python
svm_predict =svm_pipeline.predict(x_test)
```

```python
print("mae: ",mean_absolute_error(y_test,svm_predict))
print("mse: ",mean_squared_error(y_test,svm_predict))
print("r2: ",r2_score(y_test,svm_predict))
```

```python
from sklearn.neighbors import KNeighborsRegressor
```

```python
kn_pipeline = Pipeline([
    ("preprocessor",preprocessor),
    ("model",KNeighborsRegressor(n_neighbors=5))
])
```

```python
kn_pipeline.fit(x_train,y_train)
```

```python
kn_predict = kn_pipeline.predict(x_test)
```

```python
print("mae: ",mean_absolute_error(y_test,kn_predict))
print("mse: ",mean_squared_error(y_test,kn_predict))
print("r2: ",r2_score(y_test,kn_predict))
```

```python
dt_pipeline = Pipeline([
    ("preprocessor",preprocessor),
    ("model",DecisionTreeRegressor())
])
```

```python
dt_pipeline.fit(x_train,y_train)
```

```python
dt_predict = dt_pipeline.predict(x_test)
```

```python
print("mae: ",mean_absolute_error(y_test,dt_predict))
print("mse: ",mean_squared_error(y_test,dt_predict))
print("r2: ",r2_score(y_test,dt_predict))
```

```python
rf_pipeline = Pipeline([
    ("preprocessor",preprocessor),
    ("model",RandomForestRegressor())
])
```

```python
rf_pipeline.fit(x_train,y_train)
```

```python
rf_predict = rf_pipeline.predict(x_test)
```

```python
print("mae: ",mean_absolute_error(y_test,rf_predict))
print("mse: ",mean_squared_error(y_test,rf_predict))
print("r2: ",r2_score(y_test,rf_predict))
```

```python
from sklearn.ensemble import GradientBoostingRegressor
gb_pipeline = Pipeline([
    ("preprocessor",preprocessor),
    ("model",GradientBoostingRegressor())
])
```

```python
gb_pipeline.fit(x_train,y_train)
```

```python
gb_predict = gb_pipeline.predict(x_test)
```

```python
print("mae: ",mean_absolute_error(y_test,gb_predict))
print("mse: ",mean_squared_error(y_test,gb_predict))
print("r2: ",r2_score(y_test,gb_predict))
```

```python
param_grid_rf = {
    "model__n_estimators":[100,200],
    'model__max_depth':[10,20
                        ]
}
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
grid_rf = GridSearchCV(
    rf_pipeline,
    param_grid_rf,
    cv=3,
    scoring="r2",
    n_jobs=-1
)
```

```python
grid_rf.fit(x_train,y_train)
```

```python
print("best parameters: ",grid_rf.best_params_)
print("best cv score: ",grid_rf.best_score_)
```

```python
rf_predict = grid_rf.predict(x_test)
```

```python
print("mse:",mean_squared_error(y_test,rf_predict))
print("mae:",mean_absolute_error(y_test,rf_predict))
print("r2:",r2_score(y_test,rf_predict))
```
